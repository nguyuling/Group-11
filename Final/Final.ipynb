#### **basic sequence info**

```python
def retreat_prot_properties(protein_id):
    # read prot seq from fasta file
    filename = protein_id + '.fasta'
    record = SeqIO.read(filename, 'fasta')
    protein_sequence = str(record.seq)

    analysis = ProtParam.ProteinAnalysis(protein_sequence)
    properties = {
        'Protein ID' : record.id,
        'Sequence' : str(record.seq),
        'Description' : record.description,
        'Length' : f"{len(protein_sequence)} amino acids",
        'Molecular Weight' : f"{round(analysis.molecular_weight(), 4)} Da",
        'Isoelectric Point' : round(analysis.isoelectric_point(), 4),
        'Instability Index' : round(analysis.instability_index(), 4),
        'Aromaticity' : round(analysis.aromaticity(), 4),
        'Gravy' : round(analysis.gravy(), 4)
    }
    
    for key, value in properties.items():
        print(f"{key:<25} : {value}")
        
    return properties
```

```python
prot_ids = ['P01308', 'P01309', 'P01310', 'P01311']
all_properties = []

for prot_id in prot_ids:
    protein_properties = retreat_prot_properties(prot_id)
    all_properties.append(protein_properties)

df = pd.DataFrame(all_properties)
df
```

#### **DNA > RNA > Protein**

```python
# from raudhah
dna = "ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG"

# DNA TO RNA change T to U
rna = ""
for base in dna:
    if base=="T":
        rna += "U"
    else:
        rna += base

codon_table={
    "AUG":"M",
    "GCC":"A",
    "AUU":"I",
    "GUA":"V",
    "UAA":"STOP",
    "UGA":"STOP",
    "UAG":"STOP"
}

protein = ""


for i in range (0,len(rna), 3):
    codon = rna[i:i+3]
    if codon in codon_table:
        aa = codon_table[codon]
        if aa == 'STOP':
            break;
        protein += aa
        
print(f"DNA : {dna}")
print(f"RNA : {rna}")
print(f"Protein : {protein}")
```

#### **euclidean distance**

```python
def euclidean_distance(p1, p2):
    distance = 0
    for i in range (0, len(p1)):
        distance += (p1[i] - p2[i]) ** 2
    return math.sqrt(distance)
```

#### **knn**

```python
# KNN Classifier using euclidean_distance function

def knn_predict(training_data, test_point, k):

    distances = []
    
    for features, label in training_data:
        dist = euclidean_distance(features, test_point)
        distances.append(dist, label)
        
    # Sort by distance (closest first)
    distances.sort(key=lambda x: x[0])  
    neighbors = distances[:k]  

    for dist, label in neighbors:
        if label == "M":
            count_M += 1
        else:
            count_B += 1
    
    # Return majority vote
    if count_M > count_B:
        return "M"
    else:
        return "B"

# --- Example usage ---
# Define a test point (new sample to classify)
test_point = [15.0, 20.0, 100.0, 700.0, 0.1]  # Example values

# Predict using k=3
k = 3
prediction = knn_predict(data, test_point, k)

print(f"Test point: {test_point}")
print(f"K value: {k}")
print(f"Prediction: {prediction}")
print(f"  ('M' = Malignant, 'B' = Benign)")
```
