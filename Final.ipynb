#### **libraries**

```python
from Bio import Entrez, SeqIO
from Bio.SeqUtils import ProtParam
from collections import Counter
import pandas as pd
```

#### **retrieve data via `Entrez` and `SeqIO`**

```python
#! Set email (mandatory for Entrez)
Entrez.email = 'nguyuling@graduate.utm.my'
protein_id = "P01308"

#! Fetch protein
handle = Entrez.efetch(
    db='protein',
    id=protein_id,
    rettype='fasta',
    retmode='text'
)

#! Read the FASTA record from the handle
record = SeqIO.read(handle, 'fasta')
handle.close()

#! Extract the sequence string
protein_sequence = str(record.seq)

#! Save to file
filename = protein_id + ".fasta"
with open(filename, "w") as file:
    SeqIO.write(record, file, "fasta")
```

#### **basic sequence info**

```python
#! Read protein sequence from the saved FASTA file
fasta_file = "P01308.fasta"
record = SeqIO.read(fasta_file, 'fasta')
protein_sequence = str(record.seq)

analysis = ProtParam.ProteinAnalysis(protein_sequence)
properties = {
    'Protein ID' : record.id,
    'Sequence' : record.seq,
    'Description' : record.description,
    'Length' : f"{analysis.length} amino acids",
    'Molecular Weight': f"{round(analysis.molecular_weight(),4)} Da",
    'Isoelectric Point': round(analysis.isoelectric_point(),4),
    'Instability Index': round(analysis.instability_index(),4),
    'Aromaticity': round(analysis.aromaticity(),4),
    'Gravy': round(analysis.gravy(),4)
}

for key, value in properties.items():
    print(f'{key:<25}: {value}')
```

#### **percentage of each amino acid**

```python
aa_count = Counter(protein_sequence)
total = len(protein_sequence)
aa_percent = {aa: (count / total) * 100 for aa, count in aa_count.items()}

df = pd.DataFrame({
    'Amino Acid' : list(aa_count.keys()),
    'Count': list(aa_count.values()),
    'Percentage': [aa_percent[aa] for aa in aa_count.keys()]
}).sort_values(by='Count', ascending=False)

df
```
