#### **9. Chi-Squared Selection**
Use univariate feature selection with a chi-squared test is used for feature scoring to select the top five features according to the k highest scores

```python
chi2_selection = SelectKBest(score_func=chi2, k=5).fit(X, y.values.ravel())
```

#### **10. Chi-Squared: Model Building**
Build the model using fit().

```python
from sklearn.linear_model import LogisticRegression

X_selected_chi2 = chi2_selection.transform(X)
model_chi2 = LogisticRegression(max_iter=1000).fit(X_selected_chi2, y.values.ravel())
print(f"Model score: {model_chi2.score(X_selected_chi2, y.values.ravel()):.4f}")
```

#### **11. Top 5 Features with Lowest p-values**
Print the selected top five features with the lowest p-values which achieved p-values <= 0.05 using pvalues_.

```python
pvalues = chi2_selection.pvalues_

chi2_features_pvalues = pd.DataFrame({
    'features': X.columns,
    'p-value': pvalues
})
chi2_features_pvalues = chi2_features_pvalues.sort_values('p-value', ascending=True)

print("the top 5 features from p-values of chi-squared test:")
chi2_features_pvalues[chi2_features_pvalues['p-value'] <= 0.05].head(5)
```

#### **12. F-test Univariate Selection**
Use univariate feature selection with F-test is used for feature scoring to select the top five features according to the k highest scores

```python
f_selection = SelectKBest(score_func=f_classif, k=5).fit(X, y.values.ravel())
```

#### **13. F-test: Model Building**
Build the model using fit().

```python
X_selected_f = f_selection.transform(X)
model_f = LogisticRegression(max_iter=1000).fit(X_selected_f, y.values.ravel())

print(f"Model score: {model_f.score(X_selected_f, y.values.ravel()):.4f}")
```

#### **14. F-test: Top 5 Features with Lowest p-values**
Print the selected top five features with the lowest p-values which achieved p-values <= 0.05 using pvalues.

```python
pvalues = f_selection.pvalues_

f_features_pvalues = pd.DataFrame({
    'features': X.columns,
    'p-value': pvalues
})
f_features_pvalues = f_features_pvalues.sort_values('p-value', ascending=True)

print("the top 5 features from p-values of F-test:")
f_features_pvalues[f_features_pvalues['p-value'] <= 0.05].head(5)
```

#### **15. Features from P-values**
Print the same features from the p-values of the chi-squared test and F-test.

```python
chi2_significant = set(chi2_features_pvalues.head(5)['features'])
f_significant = set(f_features_pvalues.head(5)['features'])
common_significant = chi2_significant.intersection(f_significant)

print("the same features from top 5 of chi-squared test and F-test:")
common_significant
```

### **PART C - IDENTIFICATION OF TANDEM REPEATS**

#### **Task 1: Count Number of Neucleotides**

```python
def count_nucleotides(dna_sequence):
    counts = {'A': 0, 'C': 0, 'G': 0, 'T': 0}
    dna_sequence = dna_sequence.upper()
    for nucleotide in dna_sequence:
        if nucleotide in counts:
            counts[nucleotide] += 1
    return counts

# test with 3 DNA seq
part_1 = "tttaga"
part_2 = "ttcgtg"
part_3 = "ttgtga"

print(f"Part 1 Counts: {count_nucleotides(part_1)}")
print(f"Part 2 Counts: {count_nucleotides(part_2)}")
print(f"Part 3 Counts: {count_nucleotides(part_3)}")
```

#### **Task 2: Compare Microsatellite Profiles**

```python
def count_repeats(dna_sequence, motif):
    count = 0
    sequence = dna_sequence.upper()
    motif = motif.upper()
    for i in range(0, len(sequence) - len(motif) + 1, len(motif)):
        if sequence[i:i+len(motif)] == motif:
            count += 1
    return count

def compare_microsatellite_profiles(sequences, names, motif):
    repeat_counts = {}
    for name, sequence in zip(names, sequences):
        repeat_counts[name] = count_repeats(sequence, motif)
    
    max_repeats = max(repeat_counts.values())
    max_profiles = [name for name, count in repeat_counts.items() if count == max_repeats]
    
    if max_repeats == 0:
        result = f"No profile has any repeats of {motif}."
    elif len(max_profiles) > 1:
        profiles_str = ", ".join(max_profiles)
        result = f"The profiles {profiles_str} have the same most repeats of {motif}."
    else:
        result = f"The profile {max_profiles[0]} has the most repeats of {motif}."
    
    return result

# test with 3 DNA seq
part_1 = "tttaga"
part_2 = "ttcgtg"
part_3 = "ttgtga"

sequences = [part_1, part_2, part_3]
names = ["Part 1", "Part 2", "Part 3"]
motif = "TT"

# compare with microsatellite profiles
result = compare_microsatellite_profiles(sequences, names, motif)

print("Microsatellite Profile Comparison Result:")
print(result)
```
